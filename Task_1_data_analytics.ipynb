```python
import pandas as pd
import numpy as np

# TASK 1: Basic Data Operations & Cleaning
np.random.seed(101)
raw_data = {
    'Student_ID': [101, 102, 103, 104, 105, 106, 106, 107, 108, 109],
    'Age': [18, 19, np.nan, 20, 19, 18, 18, np.nan, 21, 19],
    'Study_Hours': [5.5, 4.0, 6.5, np.nan, 8.0, 3.5, 3.5, 7.0, 2.0, 9.0],
    'Attendance_Pct': [90, 85, 78, 92, np.nan, 60, 60, 88, 70, 95],
    'Grade': ['A', 'B', 'B', 'A', 'A', 'D', 'D', 'B', 'F', 'A']
}

df_task1 = pd.DataFrame(raw_data)
df_task1 = df_task1.drop_duplicates()
df_task1['Age'] = df_task1['Age'].fillna(df_task1['Age'].median())
df_task1['Study_Hours'] = df_task1['Study_Hours'].fillna(df_task1['Study_Hours'].mean())
df_task1['Attendance_Pct'] = df_task1['Attendance_Pct'].fillna(df_task1['Attendance_Pct'].mean())

print("--- TASK 1 CLEANED DATASET ---")
print(df_task1)
```
